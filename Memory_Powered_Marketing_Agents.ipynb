```python
from groq import Groq
from google.colab import userdata # Import userdata to access secrets

client = Groq(
    api_key=userdata.get("GROQ_API_KEY") # Retrieve the key from Colab secrets
)

print("Groq client connected successfully!")
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": "Explain marketing agents in one sentence."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
def ask_groq(prompt, system_prompt=None):

    messages = []

    if system_prompt:
        messages.append({
            "role": "system",
            "content": system_prompt
        })

    messages.append({
        "role": "user",
        "content": prompt
    })

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=messages,
        temperature=0.3
    )

    return response.choices[0].message.content
```

```python
result = ask_groq(
    "Give me 3 content ideas for a technology company."
)

print(result)
```

```python
def content_strategy_agent(marketing_context):

    system_prompt = """
You are the Content Strategy Agent of a marketing intelligence system.

Your job is to:
1. Analyze historical marketing information.
2. Identify successful content patterns.
3. Identify content gaps.
4. Recommend what content should be created next.

Do not invent historical performance data.
Base recommendations on the information provided.
"""

    prompt = f"""
Analyze the following marketing information:

{marketing_context}

Provide:

1. Key patterns
2. Successful topics
3. Content gaps
4. Recommended next content
5. Reason for the recommendation
"""

    return ask_groq(
        prompt,
        system_prompt
    )
```

```python
marketing_context = """
Previous content:

Post 1:
Topic: Generative AI
Format: Carousel
Likes: 1200
Comments: 180
Shares: 320

Post 2:
Topic: Cloud Computing
Format: Text
Likes: 350
Comments: 30
Shares: 20

Post 3:
Topic: AI Agents
Format: Carousel
Likes: 1500
Comments: 210
Shares: 400
"""

print(content_strategy_agent(marketing_context))
```

```python
import pandas as pd

marketing_data = pd.DataFrame([
    {
        "platform": "LinkedIn",
        "topic": "Generative AI",
        "content_type": "Carousel",
        "audience": "Developers",
        "impressions": 18430,
        "likes": 1240,
        "comments": 184,
        "shares": 320,
        "clicks": 812
    },
    {
        "platform": "LinkedIn",
        "topic": "Cloud Computing",
        "content_type": "Text",
        "audience": "Students",
        "impressions": 9200,
        "likes": 350,
        "comments": 30,
        "shares": 20,
        "clicks": 140
    },
    {
        "platform": "Instagram",
        "topic": "AI Agents",
        "content_type": "Reel",
        "audience": "Developers",
        "impressions": 25100,
        "likes": 2100,
        "comments": 260,
        "shares": 510,
        "clicks": 950
    },
    {
        "platform": "LinkedIn",
        "topic": "Machine Learning",
        "content_type": "Carousel",
        "audience": "Data Scientists",
        "impressions": 16200,
        "likes": 980,
        "comments": 120,
        "shares": 210,
        "clicks": 620
    },
    {
        "platform": "Instagram",
        "topic": "Cloud Computing",
        "content_type": "Image",
        "audience": "Students",
        "impressions": 14300,
        "likes": 620,
        "comments": 75,
        "shares": 110,
        "clicks": 280
    }
])

marketing_data
```

```python
marketing_data["engagement_rate"] = (
    marketing_data["likes"] +
    marketing_data["comments"] +
    marketing_data["shares"]
) / marketing_data["impressions"] * 100

marketing_data["ctr"] = (
    marketing_data["clicks"] /
    marketing_data["impressions"] * 100
)

marketing_data
```

```python
def get_marketing_summary(df):

    summary = {
        "total_posts": len(df),
        "platforms": df["platform"].unique().tolist(),
        "topics": df["topic"].unique().tolist(),
        "average_engagement_rate": round(
            df["engagement_rate"].mean(), 2
        ),
        "average_ctr": round(
            df["ctr"].mean(), 2
        )
    }

    return summary
```

```python
summary = get_marketing_summary(marketing_data)

summary
```

```python
def get_top_content(df, n=3):

    top = df.sort_values(
        by="engagement_rate",
        ascending=False
    ).head(n)

    return top[
        [
            "platform",
            "topic",
            "content_type",
            "audience",
            "engagement_rate",
            "ctr"
        ]
    ]
```

```python
top_content = get_top_content(marketing_data)

top_content
```

```python
def build_marketing_context(df):

    summary = get_marketing_summary(df)
    top_content = get_top_content(df)

    context = f"""
MARKETING SUMMARY

Total posts: {summary['total_posts']}

Platforms:
{summary['platforms']}

Topics:
{summary['topics']}

Average engagement rate:
{summary['average_engagement_rate']}%

Average CTR:
{summary['average_ctr']}%

TOP PERFORMING CONTENT

{top_content.to_string(index=False)}
"""

    return context
```

```python
context = build_marketing_context(marketing_data)

print(context)
```

```python
def content_strategy_agent(df):

    marketing_context = build_marketing_context(df)

    system_prompt = """
You are the Content Strategy Agent in a memory-powered
marketing intelligence system.

Your responsibilities:

1. Analyze historical marketing performance.
2. Identify successful topics and formats.
3. Identify content gaps.
4. Recommend what content should be created next.

Use the provided data as evidence.
Do not invent performance statistics.
Explain why each recommendation is made.
"""

    prompt = f"""
Here is the current marketing data:

{marketing_context}

Based on this data, provide:

1. Key performance patterns
2. Best-performing topics
3. Best-performing content formats
4. Audience insights
5. Content gaps
6. Three recommendations for future content

For every recommendation explain the evidence behind it.
"""

    return ask_groq(
        prompt,
        system_prompt
    )
```

```python
result = content_strategy_agent(marketing_data)

print(result)
```
